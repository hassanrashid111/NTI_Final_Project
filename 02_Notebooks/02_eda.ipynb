## 2️⃣ Load Data

```python
from pathlib import Path
import pandas as pd
import config
import utils

# Dictionary to hold all loaded dataframes for EDA
datasets = {}

# Mapping dataset names to config file paths
raw_files = {
    'train': (config.TRAIN_FILE, 'df_train'),
    'test': (config.TEST_FILE, 'df_test'),
    'stores': (config.STORES_FILE, 'df_stores'),
    'items': (config.ITEMS_FILE, 'df_items'),
    'oil': (config.OIL_FILE, 'df_oil'),
    'transactions': (config.TRANSACTIONS_FILE, 'df_trans'),
    'holidays_events': (config.HOLIDAYS_FILE, 'df_holidays')
}

print('📥 Loading raw dataset files for Exploratory Data Analysis...')
print(f"{'Dataset':<18} | {'Rows':<12} | {'Columns':<8} | {'Status':<10}")
print('-' * 55)

for name, (file_path, var_name) in raw_files.items():
    if file_path.exists():
        df = utils.load_csv(file_path)
        datasets[name] = df
        globals()[var_name] = df
        print(f"{name:<18} | {len(df):<12,} | {len(df.columns):<8} | ✅ Loaded")
    else:
        print(f"{name:<18} | {'N/A':<12} | {'N/A':<8} | ❌ Missing")

print('-' * 55)
print(f"✅ Successfully loaded {len(datasets)} datasets into memory. Ready for EDA!")
```
