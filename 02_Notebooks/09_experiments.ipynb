```python
# ============================================================
# Project Bootstrap (Google Colab)
# Run this cell first in every notebook.
# ============================================================
import os
import sys
from pathlib import Path

# Mount Google Drive (if running in Google Colab & not already mounted)
if 'google.colab' in sys.modules and not os.path.exists('/content/drive'):
    from google.colab import drive
    drive.mount('/content/drive')

# Project root path candidates
POSSIBLE_ROOTS = [
    Path('/content/drive/MyDrive/Colab Notebooks/Demand-Forecasting-System'),
    Path('/content/drive/MyDrive/NTI/Demand Forecasting System'),
    Path('/content/drive/MyDrive/Demand Forecasting System'),
    Path('/content/Demand Forecasting System'),
    Path.cwd(),
    Path.cwd().parent
]

PROJECT_ROOT = None
for candidate in POSSIBLE_ROOTS:
    if candidate.exists() and (candidate / 'config.py').exists():
        PROJECT_ROOT = candidate
        break

if PROJECT_ROOT is None:
    raise FileNotFoundError(
        'Project folder with config.py not found in Google Drive or local path.\n'
        'Make sure the project exists in Google Drive.'
    )

# Add project root to Python path
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

# Change working directory to project root
os.chdir(PROJECT_ROOT)

print('=' * 60)
print('Project Root :', PROJECT_ROOT)
print('Working Dir  :', os.getcwd())
print('Bootstrap completed successfully.')
print('=' * 60)

# ==============================================================================
# SECTION 1 & 2 — Imports & Load Resources
# ==============================================================================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Import central project configuration and utility functions
import config
import utils

# Random seed for reproducibility
np.random.seed(getattr(config, 'RANDOM_SEED', 42))

# Pandas display & plotting setup
pd.set_option('display.max_columns', 50)
pd.set_option('display.max_rows', 100)
pd.set_option('display.float_format', lambda x: '%.4f' % x)
sns.set_theme(style='whitegrid', palette='muted')

# 1. Load feature-engineered dataset
feature_store_path = config.FEATURE_STORE if hasattr(config, 'FEATURE_STORE') else Path('01_Dataset/features/feature_store.parquet')

print('📥 Loading feature store dataset for Research Experiments...')
if feature_store_path.exists():
    df_features = pd.read_parquet(feature_store_path) if feature_store_path.suffix == '.parquet' else pd.read_csv(feature_store_path)
    mem_bytes = df_features.memory_usage(deep=True).sum()
    mem_str = f"{mem_bytes / (1024**2):.2f} MB" if mem_bytes < 1024**3 else f"{mem_bytes / (1024**3):.2f} GB"
    print(f"  • Dataset Name : {feature_store_path.name}")
    print(f"  • Dataset Shape: {df_features.shape}")
    print(f"  • Memory Usage : {mem_str}")
else:
    print(f"⚠️ Feature store not found at {feature_store_path}.")
    df_features = None

# 2. Load trained models
print('\n📦 Scanning and loading trained models...')
loaded_models = {}
models_dir = config.MODELS_DIR if hasattr(config, 'MODELS_DIR') else Path('03_Models')

if models_dir.exists():
    for model_file in models_dir.glob('*'):
        if model_file.suffix in ['.pkl', '.joblib', '.cbm', '.pt']:
            try:
                model = utils.load_model(model_file) if hasattr(utils, 'load_model') else str(model_file)
                loaded_models[model_file.stem] = model
                print(f"  ✅ Loaded Model: {model_file.name}")
            except Exception as e:
                print(f"  ⚠️ Could not load {model_file.name}: {e}")

print('\n============================================================')
print(f"  • Loaded Models Count : {len(loaded_models)}")
print(f"  • Project Status       : ✅ Environment & Resources Ready")
print('============================================================\n')
print('✅ Research Environment Initialized. Ready for Experiments & Ablation Studies!')

# TODO: This experiment will be implemented by the Machine Learning Research team.
# raise NotImplementedError("Implemented later by the Machine Learning Research team.")
```
