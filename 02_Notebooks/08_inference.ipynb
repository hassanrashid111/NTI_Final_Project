## 🎯 Purpose & Decision Support Architecture

Translate demand forecasts into **actionable inventory procurement decisions**  
to prevent **stockouts** while avoiding **costly overstocking**.

```
Champion Model (from 03_Models/)
                ↓
16-Day Demand Forecast Generation (t + 1 → t + 16)
                ↓
┌─────────────────────────────────────────────────────────────┐
│                INVENTORY OPTIMIZATION ENGINE                │
├─────────────────────────────────────────────────────────────┤
│ 1. Safety Stock (SS) = Z × σ_d × √(Lead_Time)              │
│ 2. Reorder Point (ROP) = (Avg_Daily_Demand × Lead_Time) + SS│
│ 3. Restock Order Qty = max(0, ROP - Current_Stock)          │
└─────────────────────────────────────────────────────────────┘
                ↓
Inventory Status Alert Classification
  ├── 🔴 UNDERSTOCK Alert : Current Stock < ROP (Order Now)
  ├── 🟢 OPTIMAL Status   : Healthy Stock Level
  └── 🟡 OVERSTOCK Alert  : Current Stock > 2.5 × ROP (Reduce Order)
                ↓
Output: 01_Dataset/predictions/final_predictions.csv
```

| Specification | Detail |
|---------------|--------|
| **Inputs** | Champion model from `03_Models/` & test feature set |
| **Outputs** | `01_Dataset/predictions/final_predictions.csv` & purchase order recommendations |
