## 1️⃣ Imports & Load Best Model

````python
import pandas as pd
import numpy as np
from pathlib import Path

# Import central project configuration and utility functions
import config
import utils

# Pandas display options
pd.set_option('display.max_columns', 50)
pd.set_option('display.max_rows', 100)
pd.set_option('display.float_format', lambda x: '%.4f' % x)

# Scan and load the best model from 03_Models/
print('📦 Loading the selected Best Model for Inference...')
models_dir = config.MODELS_DIR if hasattr(config, 'MODELS_DIR') else Path('03_Models')
best_model_path = getattr(config, 'BEST_MODEL_PATH', None)

best_model = None
model_file_name = "N/A"
model_type_str = "N/A"
status_str = "❌ Model File Not Found"

if best_model_path and Path(best_model_path).exists():
    try:
        best_model = utils.load_model(best_model_path) if hasattr(utils, 'load_model') else str(best_model_path)
        model_file_name = Path(best_model_path).name
        model_type_str = type(best_model).__name__
        status_str = "✅ Loaded Successfully"
    except Exception as e:
        status_str = f"⚠️ Loading Error ({e})"
elif models_dir.exists():
    model_candidates = list(models_dir.glob('*'))
    if model_candidates:
        target_file = model_candidates[0]
        try:
            best_model = utils.load_model(target_file) if hasattr(utils, 'load_model') else str(target_file)
            model_file_name = target_file.name
            model_type_str = type(best_model).__name__
            status_str = "✅ Loaded Successfully"
        except Exception as e:
            status_str = f"⚠️ Loading Error ({e})"

print("-" * 50)
print(f"  • Selected Model : {model_file_name.split('.')[0]}")
print(f"  • Model Type     : {model_type_str}")
print(f"  • Model File     : {model_file_name}")
print(f"  • Status         : {status_str}")
print("-" * 50)
print("\n✅ Environment and Best Model loaded. Ready for Inference & Inventory Optimization!")
````
