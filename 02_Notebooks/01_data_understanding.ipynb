## 2️⃣ Load All Raw Files

Load each CSV file and store in a dictionary for easy access.

| File | Variable Name | Key Columns |
|------|---------------|-------------|
| `train.csv` | `df_train` | date, store_nbr, item_nbr, unit_sales, onpromotion |
| `stores.csv` | `df_stores` | store_nbr, city, state, type, cluster |
| `items.csv` | `df_items` | item_nbr, family, class, perishable |
| `oil.csv` | `df_oil` | date, dcoilwtico |
| `transactions.csv` | `df_trans` | date, store_nbr, transactions |
| `holidays_events.csv` | `df_holidays` | date, type, locale, description, transferred |

````python
from pathlib import Path
import pandas as pd
import config
import utils

# Dictionary to hold all loaded datasets
datasets = {}

# File definitions from config.py
raw_files = {
    'train': (config.TRAIN_FILE, 'df_train'),
    'test': (config.TEST_FILE, 'df_test'),
    'stores': (config.STORES_FILE, 'df_stores'),
    'items': (config.ITEMS_FILE, 'df_items'),
    'oil': (config.OIL_FILE, 'df_oil'),
    'transactions': (config.TRANSACTIONS_FILE, 'df_trans'),
    'holidays_events': (config.HOLIDAYS_FILE, 'df_holidays')
}

print('📥 Loading raw dataset files from 01_Dataset/raw/...')
print(f"{'Dataset':<18} | {'Rows':<12} | {'Columns':<8} | {'Memory Usage':<12}")
print('-' * 60)

for name, (file_path, var_name) in raw_files.items():
    if file_path.exists():
        df = utils.load_csv(file_path)
        datasets[name] = df
        globals()[var_name] = df
        
        mem_bytes = df.memory_usage(deep=True).sum()
        mem_str = f"{mem_bytes / (1024**2):.2f} MB" if mem_bytes < 1024**3 else f"{mem_bytes / (1024**3):.2f} GB"
        print(f"{name:<18} | {len(df):<12,} | {len(df.columns):<8} | {mem_str:<12}")
    else:
        print(f"{name:<18} | ❌ File Not Found: {file_path.name}")

print('-' * 60)
print(f"✅ Successfully loaded {len(datasets)} datasets into 'datasets' dictionary.")
````
